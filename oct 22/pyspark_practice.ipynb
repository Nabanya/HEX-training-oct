withcolumn()

```python
df = df.withColumn(
    "gross_amount",
    df.quantity * df.unit_price
)
df.show()
```

```python
df = df.withColumn(
    "discount_amount",
    df.gross_amount * df.discount_pct / 100
)

df.show()
```

```python
df = df.withColumn(
    "final_amount",
    df.gross_amount - df.discount_amount
)

df.show()
```

```python
from pyspark.sql.functions import col, when

df = df.withColumn("country", when(col("order_id").isNotNull(), "India"))

df.show()
```

```python
df = df.withColumn(
    "high_value",
    when(col("final_amount") >= 50000, "Yes")
    .otherwise("No")
)

df.show()
```

Conditional columns

```python
df = df.withColumn(
    "order_size",
    when(df.final_amount >= 50000, "High")
    .when(df.final_amount >= 15000, "Medium")
    .otherwise("Low")
)

df.show()
```

```python
df = df.withColumn(
    "quantity_type",
    when(df.quantity >= 4, "Bulk")
    .otherwise("Regular")
)

df.show()
```

```python
df = df.withColumn(
    "delivery_category",
    when(df.status == "Delivered", "Completed")
    .when(df.status.isin("Cancelled", "Returned"), "Failed")
    .otherwise("In Progress")
)

df.show()
```
