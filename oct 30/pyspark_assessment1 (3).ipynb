```python
from google.colab import files
from pyspark.sql import SparkSession, functions as F
from pyspark.sql.window import Window
from pyspark import StorageLevel

spark = SparkSession.builder.appName("Energy Analytics Assessment").master("local[*]").getOrCreate()
def load_csv(filename):
    return spark.read.csv(filename, header=True, inferSchema=True)
```

SECTION 01 / DATA PREPARATION & ANALYSIS

```python
readings_df = load_csv("meter_readings_90000.csv")
readings_extra_df = load_csv("meter_readings_extra_12000.csv")
meter_master_df = load_csv("meter_master.csv")
substation_master_df = load_csv("substation_master.csv")
tariff_master_df = load_csv("tariff_master.csv")
tickets_df = load_csv("maintenance_tickets_1600.csv")
```

```python
readings_df.printSchema()
print("Rows:", readings_df.count())
readings_df.show(10, truncate=False)
```

```python
selected_df = readings_df.select("reading_id", "meter_id", "reading_timestamp", "energy_kwh", "reading_status")
renamed_df = readings_df.withColumnRenamed("source_type", "collection_mode")
renamed_df.show(5)
```

```python
valid_high_energy_df = readings_df.filter((F.col("reading_status") == "Valid") & (F.col("energy_kwh") > 20)).orderBy(F.col("energy_kwh").desc())
valid_high_energy_df.show(10)
```

```python
energy_band_df = readings_df.withColumn("energy_band",
    F.when(F.col("energy_kwh").isNull(), "Unknown")
     .when(F.col("energy_kwh") < 5, "Low")
     .when((F.col("energy_kwh") >= 5) & (F.col("energy_kwh") <= 25), "Medium")
     .when(F.col("energy_kwh") > 25, "High").otherwise("Unknown"))
energy_band_df.show(10)
```

```python
null_counts = readings_df.select([
    F.count(F.when(F.col(c).isNull(), c)).alias(c)
    for c in ["energy_kwh", "voltage_v", "power_factor", "operator_id"]])
null_counts.show()
```

```python
filled_readings_df = readings_df.na.fill({"operator_id": "UNASSIGNED","power_factor": 0.90})
filled_readings_df.show(10)
```

```python
initial_count = readings_df.count()
cleaned_readings_df = readings_df.filter(F.col("energy_kwh").isNotNull() & F.col("voltage_v").isNotNull())
final_count = cleaned_readings_df.count()
print(f"Initial Rows: {initial_count}")
print(f"Final Rows: {final_count}")
print(f"Rows Removed: {initial_count - final_count}")
```

```python
valid_energy_by_site = readings_df.filter(F.col("reading_status") == "Valid").join(meter_master_df, "meter_id", "inner")\
.groupBy("site_type").agg(F.sum("energy_kwh").alias("total_valid_energy_kwh")).orderBy("site_type")
valid_energy_by_site.show()
```

```python
valid_energy_region_city = readings_df.filter(F.col("reading_status") == "Valid").join(meter_master_df, "meter_id", "inner")\
    .join(substation_master_df, "substation_id", "inner").groupBy("region", "city").agg(F.sum("energy_kwh").alias("total_valid_energy_kwh")).orderBy("region", "city")
valid_energy_region_city.show()
```

```python
energy_charges_df = readings_df.filter((
    F.col("reading_status") != "Rejected") & (
    F.col("energy_kwh").isNotNull())).join(meter_master_df, "meter_id", "inner").join(tariff_master_df,
    "tariff_id", "inner").withColumn("energy_charge_inr", F.col("energy_kwh") * F.col("rate_per_kwh_inr"))
energy_charges_df.select("reading_id", "meter_id", "energy_kwh", "tariff_id", "rate_per_kwh_inr", "energy_charge_inr").show(10)
```

```python
tickets_summary = tickets_df.join(meter_master_df, "meter_id", "inner").groupBy("site_type", "severity").agg(
    F.count("ticket_id").alias("ticket_count"),
    F.avg("maintenance_cost_inr").alias("avg_maintenance_cost_inr")).orderBy("site_type", "severity")
tickets_summary.show()
```

```python
substation_reliability = readings_df.join(meter_master_df, "meter_id", "inner").groupBy("substation_id").agg(
    F.avg("voltage_v").alias("avg_voltage_v"),
    F.sum("outage_minutes").alias("total_outage_minutes")).orderBy(F.col("total_outage_minutes").desc())
substation_reliability.show(5)
```

```python
status_by_source = readings_df.groupBy("source_type").pivot("reading_status", ["Valid", "Estimated", "Rejected"]).count().withColumn("total_count",
    F.coalesce(F.col("Valid"), F.lit(0)) +
    F.coalesce(F.col("Estimated"), F.lit(0)) +
    F.coalesce(F.col("Rejected"), F.lit(0))).orderBy(F.col("total_count").desc())
status_by_source.show()
```

SECTION 02 / ADVANCED DATAFRAME OPERATIONS

```python
readings_with_date = readings_df.withColumn("reading_timestamp",
    F.to_timestamp("reading_timestamp")).withColumn("year",
    F.year("reading_timestamp")).withColumn("month", F.month("reading_timestamp"))
monthly_regional_energy = readings_with_date.filter(F.col("reading_status") == "Valid") \
.join(meter_master_df, "meter_id", "inner").join(substation_master_df, "substation_id", "inner") \
.groupBy("year", "month", "region") .agg(F.sum("energy_kwh").alias("monthly_valid_energy_kwh")).orderBy("year", "month", "region")
monthly_regional_energy.show(10)
```

```python
valid_meter_energy = readings_df.filter(F.col("reading_status") == "Valid").join(
    meter_master_df, "meter_id", "inner").groupBy("substation_id", "meter_id").agg(
    F.sum("energy_kwh").alias("total_meter_energy"))
window_substation = Window.partitionBy("substation_id").orderBy(F.col("total_meter_energy").desc())
top_meters_per_substation = valid_meter_energy.withColumn("rank",
F.dense_rank().over(window_substation)).filter(F.col("rank") <= 3).orderBy("substation_id", "rank")
top_meters_per_substation.show(15)
```

```python
window_meter_time = Window.partitionBy("meter_id").orderBy(
    F.col("reading_timestamp").asc(),
    F.col("reading_id").asc()).rowsBetween(Window.unboundedPreceding, Window.currentRow)
cumulative_energy_df = readings_with_date.filter(
    F.col("reading_status") == "Valid").withColumn("cumulative_valid_energy",
    F.sum("energy_kwh").over(window_meter_time))
cumulative_energy_df.select("meter_id", "reading_timestamp", "reading_id", "energy_kwh", "cumulative_valid_energy").show(10)
```

```python
count_main = readings_df.count()
count_extra = readings_extra_df.count()
combined_readings_df = readings_df.unionByName(readings_extra_df)
count_combined = combined_readings_df.count()

print(f"Main Readings Count: {count_main}")
print(f"Extra Readings Count: {count_extra}")
print(f"Combined Union Count: {count_combined}")
```

```python
repeated_ids_df = combined_readings_df.groupBy("reading_id").count().filter(F.col("count") > 1)
repeated_ids_df.show(truncate=False)
duplicate_readings = combined_readings_df.groupBy("reading_id").count().filter(F.col("count") > 1)
print(f"Number of duplicate reading_id values: {duplicate_readings.count()}")

deduped_readings_df = combined_readings_df.dropDuplicates(["reading_id"])
print(f"Final Unique Reading Count: {deduped_readings_df.count()}")
```

```python
lazy_pipeline = readings_df.filter(
    F.col("reading_status") == "Valid").withColumn("energy_band",
    F.when(F.col("energy_kwh") > 20, "High").otherwise("Standard")).groupBy("meter_id", "energy_band").agg(
    F.avg("energy_kwh").alias("avg_energy")).orderBy(F.col("avg_energy").desc())
lazy_pipeline.explain(True)
```

```python
outage_df = readings_df.filter(F.col("outage_minutes") > 0) \
    .persist(StorageLevel.MEMORY_AND_DISK)
total_outages = outage_df.count()
avg_outage_time = outage_df.select(F.avg("outage_minutes")).collect()[0][0]
print(f"Total Outage Events: {total_outages}")
print(f"Average Outage Minutes: {avg_outage_time:.2f}")
outage_df.unpersist()
```

Final Capstone

```python
combined_df = readings_df.unionByName(readings_extra_df)
deduped_df = combined_df.dropDuplicates(["reading_id"])
```

```python
cleaned_capstone_df = deduped_df.filter(F.col("energy_kwh").isNotNull()).withColumn("is_billable",
 F.when(F.col("reading_status") != "Rejected", True).otherwise(False))
```

```python
enriched_capstone_df = cleaned_capstone_df.join(meter_master_df, "meter_id", "inner").join(
    substation_master_df, "substation_id", "inner").join(tariff_master_df, "tariff_id", "inner")
```

```python
monthly_report_df = enriched_capstone_df.withColumn("month", F.date_format(F.to_timestamp("reading_timestamp"), "yyyy-MM")
).withColumn("billable_kwh", F.when(F.col("is_billable"),
    F.col("energy_kwh")).otherwise(0.0)) .withColumn("billable_charge_inr",
    F.col("billable_kwh") * F.col("rate_per_kwh_inr")).withColumn("has_outage",
    F.when(F.col("outage_minutes") > 0, 1).otherwise(0)).groupBy("region", "site_type", "month") .agg(
        F.sum("energy_kwh").alias("total_kwh"),
        F.sum("billable_kwh").alias("total_billable_kwh"),
        F.sum("billable_charge_inr").alias("total_billable_charge_inr"),
        F.sum("outage_minutes").alias("total_outage_minutes"),
        (F.sum("has_outage") / F.count("reading_id") * 100).alias("outage_reading_percentage"))
```

```python
region_window = Window.partitionBy("region", "month").orderBy(F.col("total_billable_charge_inr").desc())
final_report_df = monthly_report_df.withColumn("site_type_rank", F.dense_rank().over(region_window)).orderBy("region", "month", "site_type_rank")
```

```python
final_report_df.persist(StorageLevel.MEMORY_AND_DISK)
print(f"Report Rows: {final_report_df.count()}")
final_report_df.show(10, truncate=False)
final_report_df.unpersist()
```

```python
capstone_output_path = "monthly_energy_reliability_report.parquet"
final_report_df.repartition(2).write.mode("overwrite").parquet(capstone_output_path)
print("Output Parquet Files:", os.listdir(capstone_output_path))
```
