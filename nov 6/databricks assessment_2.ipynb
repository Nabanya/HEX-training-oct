```python
%sql
-- Existing Registration IDs
SELECT inc.registration_id, 'Existing' AS record_type
FROM incoming_registrations inc
JOIN event_registrations_work work
    ON inc.registration_id = work.registration_id
UNION ALL
-- New Registration IDs
SELECT inc.registration_id, 'New' AS record_type
FROM incoming_registrations inc
LEFT JOIN event_registrations_work work
    ON inc.registration_id = work.registration_id
WHERE work.registration_id IS NULL;
```

```python
%sql
MERGE INTO event_registrations_work AS target
USING incoming_registrations AS source
ON target.registration_id = source.registration_id
WHEN MATCHED THEN
  UPDATE SET
    target.event_id = source.event_id,
    target.attendee_name = source.attendee_name,
    target.registration_date = source.registration_date,
    target.ticket_type = source.ticket_type,
    target.tickets = source.tickets,
    target.unit_price = source.unit_price,
    target.status = source.status,
    target.channel = source.channel
WHEN NOT MATCHED THEN
  INSERT (registration_id, event_id, attendee_name, registration_date, ticket_type, tickets, unit_price, status, channel)
  VALUES (source.registration_id, source.event_id, source.attendee_name, source.registration_date, source.ticket_type, source.tickets, source.unit_price, source.status, source.channel);
```

Parquet files vs. _delta_log directory

Parquet Files: Store actual raw underlying tabular data records in compressed columnar binary format.

_delta_log Directory: Contains transaction logs (.json commit files and .checkpoint.parquet snapshot files) recording atomic changes, metadata, schema definitions, and file add/remove actions for ACID compliance.

36. VACUUM: Cleans up tombstoned Parquet files no longer referenced by Delta log state older than retention threshold.

```python
streaming_df = (spark.readStream
                .format("rate")
                .option("rowsPerSecond", 5)
                .load())
```

```python
display(streaming_df, checkpointLocation="/Volumes/dbt/default/streaming_checkpoints/rate_stream_display")
```

```python
from pyspark.sql.functions import expr

transformed_stream = (streaming_df
                      .withColumn("transaction_id", expr("concat('TXN_', value)"))
                      .withColumn("transaction_amount", expr("value * 150")))
```

```python
filtered_stream = transformed_stream.filter(col("transaction_amount") > 500)
```

```python
from pyspark.sql.functions import when

aggregated_stream = (filtered_stream
                     .withColumn("category", when(col("transaction_amount") >= 1000, "High").otherwise("Standard"))
                     .groupBy("category")
                     .count())

query = (aggregated_stream.writeStream
         .format("memory")
         .queryName("running_category_counts")
         .outputMode("complete")
         .option("checkpointLocation", "/Volumes/dbt/default/streaming_checkpoints/running_category_counts")
         .trigger(availableNow=True)
         .start())

query.awaitTermination()
display(spark.sql("SELECT * FROM running_category_counts"))

query.stop()
```
